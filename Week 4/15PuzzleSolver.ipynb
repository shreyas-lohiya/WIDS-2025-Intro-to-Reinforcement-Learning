```python
from itertools import permutations

states_row1 = []
states_row2 = []
states_row3 = []

target1 = tuple([x for x in range(1, 5)])
target2 = tuple([x for x in range(1, 9)])
target3 = tuple([x for x in range(1, 17)])

for p1 in range(16):
    for p2 in range(16):
        for p3 in range(16):
            for p4 in range(16):
                diffs = [p1-p2, p2-p3, p3-p4, p1-p3, p2-p4, p1-p4]
                if 0 not in diffs:
                    for blank in range(16):
                        if(blank != p1 and blank != p2 and blank != p3 and blank != p4):
                            grid = [0]*16
                            grid[p1] = 1
                            grid[p2] = 2
                            grid[p3] = 3
                            grid[p4] = 4
                            grid[blank] = 16
                            states_row1.append(tuple(grid))

for p1 in range(4, 16):
    for p2 in range(4, 16):
        for p3 in range(4, 16):
            for p4 in range(4, 16):
                diffs = [p1-p2, p2-p3, p3-p4, p1-p3, p2-p4, p1-p4]
                if 0 not in diffs:
                    for blank in range(4, 16):
                        if(blank != p1 and blank != p2 and blank != p3 and blank != p4):
                            grid = [1,2,3,4] + [0]*12
                            grid[p1] = 5
                            grid[p2] = 6
                            grid[p3] = 7
                            grid[p4] = 8
                            grid[blank] = 16
                            states_row2.append(tuple(grid))

for perm in permutations([x for x in range(8, 16)]):
    grid = [x for x in range(1, 9)] + [0]*8
    for idx in range(7):
        grid[perm[idx]] = idx + 9
    grid[perm[-1]] = 16
    states_row3.append(tuple(grid))
```

```python
mdp_row1 = {s:{} for s in states_row1}
mdp_row2 = {s:{} for s in states_row2}
mdp_row3 = {s:{} for s in states_row3}

# Row 1 MDP Filling
for curr_state in mdp_row1:
    acts = ['u', 'd', 'l', 'r']
    trans = {}
    prefix = tuple(curr_state[0:4])
    for act in acts:
        next_cfg = list(curr_state).copy()
        hole_idx = next_cfg.index(16)

        if prefix == target1:
            trans[act] = [(1, curr_state, 0, False)]
            continue

        if act == 'l':
            if hole_idx % 4 == 0:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx-1] = next_cfg[hole_idx-1], next_cfg[hole_idx]

        elif act == 'r':
            if hole_idx % 4 == 3:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx+1] = next_cfg[hole_idx+1], next_cfg[hole_idx]

        elif act == 'u':
            if hole_idx // 4 == 0:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx-4] = next_cfg[hole_idx-4], next_cfg[hole_idx]

        elif act == 'd':
            if hole_idx // 4 == 3:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx+4] = next_cfg[hole_idx+4], next_cfg[hole_idx]

        correct = 0
        for i in range(4):
            if next_cfg[i] == target1[i]:
                correct += 1

        reward = -2 + 0.5 * correct
        if next_cfg[0:4] == list(target1):
            reward = 67

        trans[act] = [(1, tuple(next_cfg), reward, reward == 1)]
    mdp_row1[curr_state] = trans

# Row 2 MDP Filling
for curr_state in mdp_row2:
    acts = ['u', 'd', 'l', 'r']
    trans = {}
    prefix = tuple(curr_state[0:8])
    for act in acts:
        next_cfg = list(curr_state).copy()
        hole_idx = next_cfg.index(16)

        if prefix == target2:
            trans[act] = [(1, curr_state, 0, False)]
            continue

        if act == 'l':
            if hole_idx % 4 == 0:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx-1] = next_cfg[hole_idx-1], next_cfg[hole_idx]

        elif act == 'r':
            if hole_idx % 4 == 3:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx+1] = next_cfg[hole_idx+1], next_cfg[hole_idx]

        elif act == 'u':
            if hole_idx // 4 <= 1:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx-4] = next_cfg[hole_idx-4], next_cfg[hole_idx]

        elif act == 'd':
            if hole_idx // 4 == 3:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx+4] = next_cfg[hole_idx+4], next_cfg[hole_idx]

        correct = 0
        for i in range(8):
            if next_cfg[i] == target2[i]:
                correct += 1

        reward = -5 + 0.5 * correct
        if next_cfg[0:8] == list(target2):
            reward = 140

        trans[act] = [(1, tuple(next_cfg), reward, reward == 1)]
    mdp_row2[curr_state] = trans

# Row 3 MDP Filling
for curr_state in mdp_row3:
    acts = ['u', 'd', 'l', 'r']
    trans = {}
    prefix = tuple(curr_state[0:16])
    for act in acts:
        next_cfg = list(curr_state).copy()
        hole_idx = next_cfg.index(16)

        if prefix == target3:
            trans[act] = [(1, curr_state, 0, False)]
            continue

        if act == 'l':
            if hole_idx % 4 == 0:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx-1] = next_cfg[hole_idx-1], next_cfg[hole_idx]

        elif act == 'r':
            if hole_idx % 4 == 3:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx+1] = next_cfg[hole_idx+1], next_cfg[hole_idx]

        elif act == 'u':
            if hole_idx // 4 <= 2:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx-4] = next_cfg[hole_idx-4], next_cfg[hole_idx]

        elif act == 'd':
            if hole_idx // 4 == 3:
                trans[act] = [(1, tuple(next_cfg), -67, False)]
                continue
            next_cfg[hole_idx], next_cfg[hole_idx+4] = next_cfg[hole_idx+4], next_cfg[hole_idx]

        correct = 0
        for i in range(16):
            if next_cfg[i] == target3[i]:
                correct += 1

        reward = -8 + 0.5 * correct
        if next_cfg == list(target3):
            reward = 250

        trans[act] = [(1, tuple(next_cfg), reward, reward == 1)]
    mdp_row3[curr_state] = trans
```

```python
def policy_iteration(state_space, target, mdp_dict, prefix_len, discount):
    action_set = ['u', 'd', 'r', 'l']
    value_fn = {}
    updated_value_fn = {}
    prev_policy = {}
    curr_policy = {}

    for s in state_space:
        value_fn[s] = 10.0
        updated_value_fn[s] = 0.0
        prev_policy[s] = 'u'
        curr_policy[s] = 'd'
        if target == s[:prefix_len]:
            value_fn[s] = 1
            updated_value_fn[s] = 1

    epsilon = 1e-4

    while curr_policy != prev_policy:
        prev_policy = curr_policy.copy()

        # Policy Evaluation
        while True:
            max_delta = 0
            value_fn = updated_value_fn.copy()

            for s in state_space:
                if list(s)[0:prefix_len] == target:
                    continue
                total = 0
                transitions = mdp_dict[s][curr_policy[s]]
                for t in transitions:
                    total += t[0] * (t[2] + discount * value_fn[t[1]] * (t[3] == False))
                updated_value_fn[s] = total
                max_delta = max(max_delta, abs(updated_value_fn[s] - value_fn[s]))

            if max_delta < epsilon:
                break

        # Policy Improvement
        for s in state_space:
            if list(s)[0:prefix_len] == target:
                continue
            best_action = 'u'
            best_value = -float('inf')

            for act in action_set:
                expected = 0
                transitions = mdp_dict[s][act]
                for t in transitions:
                    expected += t[0] * (t[2] + discount * updated_value_fn[t[1]] * (t[3] == False))
                if expected > best_value:
                    best_action = act
                    best_value = expected

            curr_policy[s] = best_action

    return curr_policy
```

```python
def value_iteration(state_space, target, mdp_dict, prefix_len, discount):
    action_set = ['u', 'd', 'r', 'l']
    value_fn = {}
    updated_value_fn = {}
    prev_policy = {}
    curr_policy = {}

    for s in state_space:
        value_fn[s] = -10.0
        updated_value_fn[s] = 0.0
        prev_policy[s] = 'u'
        curr_policy[s] = 'd'
        if target == s[:prefix_len]:
            value_fn[s] = 1
            updated_value_fn[s] = 1

    epsilon = 1e-2

    while True:
        prev_policy = curr_policy.copy()
        value_fn = updated_value_fn.copy()
        delta = 0

        for s in state_space:
            if list(s)[0:prefix_len] == target:
                continue
            best_val = -float('inf')
            best_act = None

            for act in action_set:
                t = mdp_dict[s][act][0]
                val = t[0] * (t[2] + discount * value_fn[t[1]])
                if val > best_val:
                    best_val = val
                    best_act = act

            updated_value_fn[s] = best_val
            delta = max(delta, abs(updated_value_fn[s] - value_fn[s]))

        if delta < epsilon:
            for s in state_space:
                if list(s)[0:prefix_len] == target:
                    continue
                best_val = -float('inf')
                best_act = None

                for act in action_set:
                    t = mdp_dict[s][act][0]
                    val = t[0] * (t[2] + discount * updated_value_fn[t[1]])
                    if val > best_val:
                        best_val = val
                        best_act = act

                curr_policy[s] = best_act
            break

    return curr_policy
```

```python
policy_row1 = value_iteration(states_row1, target1, mdp_row1, 4, 0.9)
policy_row2 = value_iteration(states_row2, target2, mdp_row2, 8, 0.9)
policy_row3 = value_iteration(states_row3, target3, mdp_row3, 16, 0.9)


def next_state_func(curr_state, action):
    new_state = list(curr_state).copy()
    empty = new_state.index(16)

    if action == 'l':
        if empty % 4 == 0:
            return curr_state
        new_state[empty], new_state[empty-1] = new_state[empty-1], new_state[empty]

    elif action == 'r':
        if empty % 4 == 3:
            return curr_state
        new_state[empty], new_state[empty+1] = new_state[empty+1], new_state[empty]

    elif action == 'u':
        if empty // 4 == 0:
            return curr_state
        new_state[empty], new_state[empty-4] = new_state[empty-4], new_state[empty]

    elif action == 'd':
        if empty // 4 == 3:
            return curr_state
        new_state[empty], new_state[empty+4] = new_state[empty+4], new_state[empty]

    return tuple(new_state)


def state_filter(state, limit):
    filtered = list(state).copy()
    for i in range(16):
        if filtered[i] > limit and filtered[i] != 16:
            filtered[i] = 0
    return tuple(filtered)


def display_state(state):
    line = "+----+----+----+----+"
    print(line)
    for i in range(0, 16, 4):
        row = [
            "" if x == 16 else x
            for x in state[i:i+4]
        ]
        print("|{:^4}|{:^4}|{:^4}|{:^4}|".format(*row))
        print(line)
```

```python
starting_state = (
    11, 15, 12, 3,
    2, 13, 10, 5,
    1, 7, 4, 8,
    6, 16, 9, 14
)

direction_map = {
    'u': 'Up',
    'd': 'Down',
    'l': 'Left',
    'r': 'Right'
}

move_count = 0
current_state = tuple(list(starting_state).copy())

display_state(starting_state)

while current_state[0:4] != target1:
    act = policy_row1[state_filter(current_state, 4)]
    move_count += 1
    print(f"move {move_count} {direction_map[act]}")
    current_state = next_state_func(current_state, act)
    display_state(current_state)
print("ROW 1 DONE")

while current_state[0:8] != target2:
    act = policy_row2[state_filter(current_state, 8)]
    move_count += 1
    print(f"move {move_count} {direction_map[act]}")
    current_state = next_state_func(current_state, act)
    display_state(current_state)
print("ROW 2 DONE")

while current_state != target3:
    act = policy_row3[state_filter(current_state, 16)]
    move_count += 1
    print(f"move {move_count} {direction_map[act]}")
    current_state = next_state_func(current_state, act)
    display_state(current_state)

print("PUZZLE DONE IN", move_count, "MOVES")
```
